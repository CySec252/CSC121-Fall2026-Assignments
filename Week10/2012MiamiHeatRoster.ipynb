```python
# NAME: roster.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

def add_players(player_list):
    """
    Accepts a list as a parameter and populates it with player names
    entered by the user until 'quit' is typed.
    """
    # 1. Start an indefinite while loop
    while True:
        # 2. Prompt user to enter a player name or type 'quit' to stop
        name = input("Enter player name (or 'quit' to stop): ").strip()

        # 3. Check if the entered text equals "quit"
        if name.lower() == "quit":
            # a. If it equals "quit", break out of the loop
            break

        # Ignore accidental empty inputs before appending
        if name:
            # b. Otherwise, add the player name to player_list using .append()
            player_list.append(name)


def is_on_roster(player_list, player_name):
    """
    Fruitful function that accepts a list and a target name.
    Returns True if the player is in the list, False otherwise.
    """
    # 1. Iterate through each name in player_list using a for loop
    for player in player_list:
        # 2. Check if current name matches the target player_name (case-insensitive)
        if player.lower() == player_name.strip().lower():
            # a. Match found: return True
            return True

    # 3. Loop completed without match: return False
    return False


# PROGRAM LOGIC

# 1. Create an empty list called roster
roster = []

# 2. Call add_players to fill the list
add_players(roster)

# 3. Ask the user for one specific name to search for
search_target = input("Who would you like to search for? ").strip()

# 4. Call is_on_roster and evaluate the returned Boolean
if is_on_roster(roster, search_target):
    # 5a. Result is True
    print("Result: Player found!")
else:
    # 5b. Result is False
    print("Result: Player not found.")
```
