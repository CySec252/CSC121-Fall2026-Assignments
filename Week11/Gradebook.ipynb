```python
# NAME: gradebook.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

def add_students(student_dict):
    # Step 1: Start an infinite loop to collect student data repeatedly
    while True:
        # Step 2: Prompt user for student name; strip() removes accidental whitespace
        name = input("Enter student name (or 'done'): ").strip()

        # Step 3: Check exit condition (case-insensitive check for 'done')
        if name.lower() == "done":
            break  # Exit the while loop

        # Step 4: Prompt for the score and convert the string input to a float
        grade = float(input(f"Enter {name}'s grade: "))

        # Step 5: Assign the student name as the key and their grade as the value
        student_dict[name] = grade


def calculate_average(student_dict):
    # Step 1: Extract all numerical grades from the dictionary using .values()
    scores = student_dict.values()

    # Step 2: Guard against ZeroDivisionError if the dictionary is empty
    if not scores:
        return 0.0

    # Step 3: Calculate the arithmetic mean: total score divided by count of entries
    average = sum(scores) / len(student_dict)

    # Step 4: Return the computed average
    return average


# PROGRAM LOGIC

# Step 1: Initialize an empty dictionary to store student records
gradebook = {}

# Step 2: Populate the dictionary using the add_students function
add_students(gradebook)

# Step 3: Print the header for the final report
print("\n--- Class Report ---")

# Step 4: Verify whether any records were entered before generating stats
if not gradebook:
    print("No students were entered.")
else:
    # Step 5: Iterate through key-value pairs using .items() to print each student
    for student, score in gradebook.items():
        # Display as a clean whole number if score is an integer (e.g., 95 instead of 95.0)
        if score.is_integer():
            print(f"{student}: {int(score)}")
        else:
            print(f"{student}: {score}")

    # Step 6: Call calculate_average to get the final score mean
    avg = calculate_average(gradebook)

    # Step 7: Print the class average formatted to two decimal places using :.2f
    print(f"Class Average: {avg:.2f}")
```
