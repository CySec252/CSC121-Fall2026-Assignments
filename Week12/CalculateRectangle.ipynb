```python
# NAME: rectangle.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# Define the function to accept length and width as parameters
def calculate_rectangle(length, width):
    # Compute the area of the rectangle
    area = length * width
    # Compute the perimeter of the rectangle
    perimeter = 2 * (length + width)
    # Pack both calculated values into a single tuple and return it
    return (area, perimeter)

# PROGRAM LOGIC
# Prompt the user for the rectangle's length and convert the string input to a float
user_length = float(input("Enter length: "))
# Prompt the user for the rectangle's width and convert the string input to a float
user_width = float(input("Enter width: "))

# Call the function with user inputs and unpack the returned tuple into two variables
my_area, my_perimeter = calculate_rectangle(user_length, user_width)

# Print a visual divider header for the output section
print("--- Results ---")
# Display the unpacked area value
print(f"The Area is: {my_area}")
# Display the unpacked perimeter value
print(f"The Perimeter is: {my_perimeter}")
```
