```python
# NAME: journal.py

# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
def write_entry(filename, entry):
    file = open(filename, 'a')
    file.write(entry + "\n")
    file.close()

def read_journal(filename):
    try:
        file = open(filename, 'r')
        contents = file.read()
        file.close()

        print("\n--- Your Journal ---")
        # .strip() removes the trailing blank line at the very end for cleaner output
        print(contents.strip())
    except FileNotFoundError:
        print("\nNo journal found yet! Select 'W' to write your first entry.")

# PROGRAM LOGIC
filename = "journal.txt"

while True:
    choice = input("\n(W)rite, (R)ead, or (Q)uit? ").strip().upper()

    if choice == 'W':
        entry = input("Enter your entry: ")
        write_entry(filename, entry)
        print("Entry saved.")
    elif choice == 'R':
        read_journal(filename)
    elif choice == 'Q' or choice == 'QUIT':
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Please enter W, R, or Q.")
```
