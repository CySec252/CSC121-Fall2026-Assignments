```python
# NAME: rentalcar.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

def validate_classification():
    """
    Prompt the user repeatedly until a valid car code ('B', 'S', 'L') is entered.
    Handles case-insensitivity by converting input to uppercase.
    """
    valid_codes = ("B", "S", "L")
    while True:
        code = input("Enter classification code (B, S, L): ").strip().upper()
        if code in valid_codes:
            return code
        print("Invalid selection. Please enter 'B', 'S', or 'L'.")


def calculate_mileage(start_odo, end_odo):
    """
    Calculate total miles driven and billable miles over the 100-mile allowance.
    """
    total_miles = end_odo - start_odo
    if total_miles > 100:
        charged_miles = total_miles - 100
    else:
        charged_miles = 0
    return total_miles, charged_miles


def calculate_total_bill(classification, days, charged_miles):
    """
    Fruitful function: Computes the rental cost.
    Strictly contains NO print statements; returns calculated float to caller.
    """
    # Daily rate mapping
    rates = {
        "B": 40.0,
        "S": 60.0,
        "L": 80.0
    }

    daily_rate = rates[classification]
    base_charge = days * daily_rate
    mileage_charge = charged_miles * 0.35

    total = base_charge + mileage_charge
    return total


# PROGRAM LOGIC

def main():
    # Classification name mapping for final presentation
    class_names = {
        "B": "Budget",
        "S": "Standard",
        "L": "Luxury"
    }

    # 1. Get validated inputs
    code = validate_classification()
    start_odo = float(input("Start Odometer: "))
    end_odo = float(input("End Odometer: "))
    days = int(input("Days Rented: "))

    # 2. Mileage logic
    total_miles, charged_miles = calculate_mileage(start_odo, end_odo)

    # 3. Calculate bill (catching the returned value from fruitful function)
    final_bill = calculate_total_bill(code, days, charged_miles)

    # 4. Display formatted summary report
    print("\n--- RENTAL SUMMARY ---")
    print(f"Classification: {class_names[code]}")
    # Display miles cleanly as integers if whole numbers
    print(f"Total Miles: {int(total_miles) if total_miles.is_integer() else total_miles}")
    print(f"Charged Miles: {int(charged_miles) if charged_miles.is_integer() else charged_miles}")
    print(f"Total Bill: ${final_bill:g}")

if __name__ == "__main__":
    main()
```
