```python
# IMPORT STATEMENTS
# (None)

# FUNCTION DEFINITIONS
# 1. Define validate_classification with one parameter: code
#    - Check if code is 'B', 'S', or 'L'
#    - Return True if valid, False otherwise
def validate_classification(code):
    clean_code = code.strip().upper()
    return clean_code in ["B", "S", "L"]


# 2. Define calculate_mileage_cost with one parameter: miles
#    - First 100 miles are free
#    - If miles > 100, calculate extra miles at $0.35 per mile
#    - Return the mileage cost (no print statement)
def calculate_mileage_cost(miles):
    if miles > 100:
        extra_miles = miles - 100
        return extra_miles * 0.35
    return 0.0


# 3. Define calculate_rental_cost with three parameters: code, days, miles
#    - Set daily rate based on code: B = $40, S = $60, L = $80
#    - Base cost is daily rate multiplied by days
#    - Call calculate_mileage_cost to get mileage fee
#    - Return total bill (base cost + mileage fee, no print statement)
def calculate_rental_cost(code, days, miles):
    clean_code = code.strip().upper()

    if clean_code == "B":
        daily_rate = 40.0
    elif clean_code == "S":
        daily_rate = 60.0
    else:  # 'L'
        daily_rate = 80.0

    base_cost = daily_rate * days
    mileage_cost = calculate_mileage_cost(miles)

    return base_cost + mileage_cost


# PROGRAM LOGIC
# 1. Prompt user for classification code (B, S, L) using a while loop for validation
code = input("Enter car classification code (B, S, L): ")
while not validate_classification(code):
    print("Invalid code. Please enter 'B', 'S', or 'L'.")
    code = input("Enter car classification code (B, S, L): ")
code = code.strip().upper()

# 2. Prompt user for number of days rented
days = int(input("Enter number of rental days: "))

# 3. Prompt user for odometer readings and calculate total miles driven
start_odometer = float(input("Enter starting odometer reading: "))
end_odometer = float(input("Enter ending odometer reading: "))
miles = end_odometer - start_odometer

# 4. Determine full classification name
if code == "B":
    class_name = "Budget"
elif code == "S":
    class_name = "Standard"
else:
    class_name = "Luxury"

# 5. Calculate chargeable miles over 100
if miles > 100:
    chargeable_miles = miles - 100
else:
    chargeable_miles = 0.0

# 6. Call calculate_rental_cost to get total bill
total_bill = calculate_rental_cost(code, days, miles)

# Commit 3 will complete the final formatted receipt output
```
