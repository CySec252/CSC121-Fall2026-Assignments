```python
# IMPORT STATEMENTS
import math

# FUNCTION DEFINITIONS
def area(radius):
    """Calculates and returns the area of a circle: pi * r^2."""
    return math.pi * (radius ** 2)

def circumference(radius):
    """Calculates and returns the circumference of a circle: 2 * pi * r."""
    return 2 * math.pi * radius

def volume(radius, height):
    """
    Calculates cylinder volume using the fruitful approach:
    calls area(radius) for base area, then multiplies by height.
    """
    base_area = area(radius)
    return base_area * height

# PROGRAM LOGIC
# 1. Input for radius
r = float(input("Enter radius: "))

# 2. Call and print area and circumference
print(f"Area: {area(r)}")
print(f"Circumference: {circumference(r)}")

# 3. Input for height
h = float(input("Enter height: "))

# 4. Call and print cylinder volume
print(f"Cylinder Volume: {volume(r, h)}")
```
