```python
# IMPORT STATEMENTS
import math

# FUNCTION DEFINITIONS
def area(radius):
    """Calculates and returns the area of a circle: pi * r^2."""
    return math.pi * (radius ** 2)

def circumference(radius):
    """Calculates and returns the circumference of a circle: 2 * pi * r."""
    return 2 * math.pi * radius

# 3. Define volume(radius, height):
#    - Call area(radius) and multiply by height.
#    - Return the result.

# PROGRAM LOGIC
# Collect radius input
r = float(input("Enter radius: "))

# Call and print area and circumference
print(f"Area: {area(r)}")
print(f"Circumference: {circumference(r)}")

# 3. Input for height (float)
# 4. Call and print volume.
```
