```python
# IMPORT STATEMENTS
# (None)


# FUNCTION DEFINITIONS
def verify_pin():
    while True:
        entered_pin = input("Enter PIN: ")
        if entered_pin == "1234":
            print("PIN Verified.")
            return True
        else:
            print("Incorrect PIN. Try again.")


def withdraw(balance, amount):
    if amount <= balance:
        new_balance = balance - amount
        print("Withdrawal successful.")
        return new_balance
    else:
        print("Insufficient funds")
        return balance


# PROGRAM LOGIC
current_balance = 500.0

verify_pin()

amount_input = float(input("Enter withdrawal amount: "))
current_balance = withdraw(current_balance, amount_input)

print(f"Current Balance: ${current_balance}")
```
