```python
# IMPORT STATEMENTS
# No external libraries needed for this assignment

# FUNCTION DEFINITIONS
def is_palindrome(word):
    """Check if a word reads the same forward and backward."""
    clean_word = word.lower()
    return clean_word == clean_word[::-1]

# PROGRAM LOGIC
user_word = input("Enter a word to test: ")

if is_palindrome(user_word):
    print(f"'{user_word}' is a palindrome!")
else:
    print(f"'{user_word}' is not a palindrome.")
```
