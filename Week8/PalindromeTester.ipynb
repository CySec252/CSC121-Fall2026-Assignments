```python
# IMPORT STATEMENTS
# No external libraries needed for this assignment

# FUNCTION DEFINITIONS
def is_palindrome(word):
    """Check if a word reads the same forward and backward."""
    clean_word = word.lower()
    reversed_word = clean_word[::-1]
    return clean_word == reversed_word

# PROGRAM LOGIC
# Temporary test calls to verify Boolean returns
print("Test 'Bob':", is_palindrome("Bob"))        # Expected: True
print("Test 'Noon':", is_palindrome("Noon"))      # Expected: True
print("Test 'Python':", is_palindrome("Python"))  # Expected: False
```
